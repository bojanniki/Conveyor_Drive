# Conveyor Drive Project

### Zadane veličine:


* **Snaga (P):** 250 W
* **Ulazna brzina ($n_1$):** 1800 RPM
* **Izlazna brzina ($n_2$):** 450 RPM (4:1 Redukcija)
* **Konfiguracija:** Paralelne osovine

## Formule i izračun

### 1. Ulazni parametri (Specifikacije pogona)
Ovi podaci definiraju radne uvjete stroja.

| Varijabla | Naziv | Opis |
| :--- | :--- | :--- |
| $P$ | **Snaga** [W] | Ukupni rad motora u jedinici vremena. |
| $n_1, n_2$ | **Brzina vrtnje** [RPM] | Broj okretaja ulaznog i izlaznog vratila u minuti. |
| $\eta$ | **Efikasnost** | Faktor gubitaka (trenje u zahvatu i ležajevima). |
| $\alpha$ | **Kut zahvata** [°] | Standardni kut pod kojim zubi pritišću (obično 20°). |

---

### 2. Kinematika i Momenti
Transformacija gibanja i sila unutar prijenosnika.

* **Prijenosni omjer ($i$):** Odnos brzina i momenata.
  $$i = \frac{n_1}{n_2}$$
* **Kutna brzina ($\omega$):** Brzina rotacije u radijanima po sekundi.
  $$\omega = \frac{2 \cdot \pi \cdot n}{60}$$
* **Okretni moment ($T$):** Zakretna sila na vratilu.
  $$T = \frac{P}{\omega}$$

---

### 3. Geometrija zupčanika
Dimenzije potrebne za 3D modeliranje i izradu.

* **Modul ($m$):** Osnovna jedinica veličine zuba (standardizirana).
* **Diobeni promjer ($d$):** Promjer na kojem se zupčanici teoretski kotrljaju.
  $$d = m \cdot z$$
* **Tjemeni ($d_a$) i Podnožni ($d_f$) promjer:** Vanjske i unutarnje granice zupčanika.
* **Osni razmak ($a$):** Razmak između središta dva vratila.
  $$a = \frac{d_1 + d_2}{2}$$
* **Obodna brzina ($v$):** Brzina gibanja zubi u točki zahvata (bitno za podmazivanje).

---

### 4. Sile u zahvatu
Mehanička opterećenja koja djeluju na zube i ležajeve.



* **Obodna sila ($F_t$):** Glavna korisna sila koja prenosi moment i savija zub.
* **Radijalna sila ($F_r$):** Sila koja pritišće vratila prema van (rezultat kuta $\alpha$).

---

### 5. Analiza čvrstoće i kvalitete
Inženjerska provjera sigurnosti i dugovječnosti.

* **Faktor prekrivanja ($\epsilon_\alpha$):** Broj parova zubi u istovremenom dodiru. Osigurava tihi rad.
* **Lewisov faktor oblika ($Y_f$):** Koeficijent koji uzima u obzir geometrijsku slabost baze zuba.
* **Napon u korijenu ($\sigma_F$):** Stvarni napon savijanja na dnu zuba.
  $$\sigma_F = \frac{F_t}{b \cdot m \cdot Y_F}$$
* **Dodirni napon ($\sigma_H$):** Hertzov tlak na bokovima zuba. Određuje otpornost na *pitting*.
  $$\sigma_H = Z_E \cdot Z_H \cdot \sqrt{\frac{F_t}{b \cdot d_1} \cdot \frac{i+1}{i}}$$



---

### 6. Faktori sigurnosti ($S_F, S_H$)
Konačni sud o dizajnu. Omjer dopuštenog naprezanja materijala i stvarnog izračunatog naprezanja. Ciljane vrijednosti za industrijske strojeve su obično $S > 1.2$.

### Interpretacija rezultata čvrstoće i faktora sigurnosti

Nakon provedenog numeričkog izračuna, dobivene su vrijednosti naprezanja koje definiraju pouzdanost reduktora.

---

### 1. Sigurnost na lom zuba ($S_F = 13.74$)
Izračunata vrijednost napona u korijenu iznosi **14.55 N/mm²**, što rezultira visokim faktorom sigurnosti u odnosu na dopuštenih 200 N/mm².

* **Analiza:** Ova visoka sigurnost je uobičajena za manje snage (250W) i robusne module (m=1.5). 
* **Inženjerski zaključak:** Zupčanik je imun na iznenadne lomove korijena zuba. Ovako visoka rezerva služi za amortizaciju ekstremnih dinamičkih udara i startnih momenata motora koji mogu biti višestruko veći od nazivnih.



---

### 2. Sigurnost na pitting/trošenje ($S_H = 2.17$)
Hertzov dodirni tlak na bokovima zuba iznosi **253.00 N/mm²**. 

* **Analiza:** Iako je ovaj napon brojčano puno veći od napona u korijenu, on je i dalje unutar granica izdržljivosti materijala C45 ($\sigma_{H, dop} = 550$ N/mm²).
* **Mjerodavnost:** Primjećuje se da je $S_H$ (2.17) značajno manji od $S_F$ (13.74). To potvrđuje osnovno pravilo strojarstva: **Zupčanici su gotovo uvijek limitirani površinskom izdržljivošću (trošenjem), a ne čvrstoćom na lom.**



---

### 3. Konačna ocjena dizajna
Dizajn se smatra **optimalnim i sigurnim**. 

1. **Dugovječnost:** Faktor sigurnosti na pitting od preko 2.0 jamči dug radni vijek (iznad 10.000 radnih sati) uz pravilno podmazivanje.
2. **Geometrijska stabilnost:** Modul 1.5 i širina 15 mm pružaju dovoljnu krutost da ne dolazi do velikih deformacija zuba pod opterećenjem, što održava točnost prijenosnog omjera.


```python
import math

# ==========================================
# 1. ULAZNI PODACI (Zadane veličine)
# ==========================================
P = 250              # Snaga [W]
n1 = 1800            # Ulazna brzina [RPM]
n2 = 450             # Izlazna brzina [RPM]
eta = 0.97           # Efikasnost (gubici trenja)
alpha_deg = 20       # Kut zahvata [stupnjevi]

# Geometrijski izbor
m = 1.5              # Modul [mm]
z1 = 18              # Broj zuba piniona

# Materijal i faktori (Čelik C45)
b = 15.0             # Širina zupčanika [mm] (fiksno za proračun napona)
Yf = 0.30            # Lewisov faktor oblika (za z=18)
ZE = 190.0           # Faktor elastičnosti (za čelik/čelik [sqrt(N/mm2)])
sigma_F_dop = 200    # Dopušteni napon savijanja [N/mm2]
sigma_H_dop = 550    # Dopušteni dodirni tlak [N/mm2]

# ==========================================
# 2. KINEMATIKA I MOMENTI
# ==========================================
i = n1 / n2
omega1 = (2 * math.pi * n1) / 60
T1 = P / omega1
T2 = T1 * i * eta

# ==========================================
# 3. GEOMETRIJA I SILA
# ==========================================
z2 = int(z1 * i)
d1 = m * z1
d2 = m * z2
da1, da2 = d1 + 2*m, d2 + 2*m
df1, df2 = d1 - 2.5*m, d2 - 2.5*m
a = (d1 + d2) / 2
v = (d1 * math.pi * n1) / 60000

# Sile
Ft = (2000 * T1) / d1
alpha_rad = math.radians(alpha_deg)
Fr = Ft * math.tan(alpha_rad)

# ==========================================
# 4. FAKTOR PREKRIVANJA (Contact Ratio)
# ==========================================
# Osnovni polumjeri (rb = r * cos(alpha))
rb1 = (d1 / 2) * math.cos(alpha_rad)
rb2 = (d2 / 2) * math.cos(alpha_rad)

# Tjemeni polumjeri (ra = da / 2)
ra1, ra2 = da1 / 2, da2 / 2

# Duljina zahvatne crte (g_alpha)
g_alpha = (math.sqrt(ra1**2 - rb1**2) + 
           math.sqrt(ra2**2 - rb2**2) - 
           a * math.sin(alpha_rad))

# Osnovni korak (pb)
pb = m * math.pi * math.cos(alpha_rad)
epsilon = g_alpha / pb

# ==========================================
# 5. PRORAČUN NAPONA (Sigma F i Sigma H)
# ==========================================
# Napon u korijenu (Savijanje)
sigma_F = Ft / (b * m * Yf)

# Dodirni napon (Hertzov tlak)
# ZH - faktor zahvata
ZH = math.sqrt(2 * math.cos(alpha_rad) / (math.sin(alpha_rad) * math.cos(alpha_rad)))
sigma_H = ZE * ZH * math.sqrt((Ft / (b * d1)) * ((i + 1) / i))

# ==========================================
# 6. INŽENJERSKI IZVJEŠTAJ
# ==========================================
print(f"{' REZULTATI PRORAČUNA REDUKTORA ':=^50}")
print(f"Prijenosni omjer (i):           {i:.2f}")
print(f"Izlazni moment (T2):            {T2:.2f} Nm")
print(f"Obodna brzina (v):              {v:.2f} m/s")
print("-" * 50)
print(f"BROJ ZUBA:                      z1 = {z1}, z2 = {z2}")
print(f"OSNI RAZMAK (a):                {a:.2f} mm")
print(f"ŠIRINA ZUPČANIKA (b):           {b:.1f} mm")
print(f"FAKTOR PREKRIVANJA (eps):       {epsilon:.3f}")
print("-" * 50)
print(f"OBODNA SILA (Ft):               {Ft:.2f} N")
print(f"RADIJALNA SILA (Fr):            {Fr:.2f} N")
print("-" * 50)
print(f"NAPON U KORIJENU (sigma_F):     {sigma_F:.2f} N/mm2")
print(f"Sigurnost na lom (SF):          {sigma_F_dop/sigma_F:.2f}")
print("-" * 50)
print(f"DODIRNI NAPON (sigma_H):        {sigma_H:.2f} N/mm2")
print(f"Sigurnost na pitting (SH):      {sigma_H_dop/sigma_H:.2f}")
print("=" * 50)

# Alarmi i upozorenja
if z1 < 17:
    print("WARNING: Rizik od podrezivanja (z1 < 17)!")
if epsilon < 1.2:
    print("WARNING: Niski faktor prekrivanja (Buka/Udarci)!")
if sigma_F > sigma_F_dop:
    print("DANGER: Previsok napon u korijenu! Puknuće zuba!")
if sigma_H > sigma_H_dop:
    print("DANGER: Previsok dodirni tlak! Pojava pittinga!")
```

## Zaključci

### 1. Kinematički Integritet i Prijenosni Omjer
* **Zaključak:** Postignut precizan prijenosni omjer  koristeći omjer zubi .
* **Inženjersko opravdanje:** Odabir  je strateški. Iako je teoretski minimum 17, 18 zubi pruža dodatnu sigurnost od **podrezivanja (undercutting)** i omogućuje korištenje standardnih alata za izradu bez potrebe za pomakom profila (x=0).

### 2. Geometrijska Optimizacija (Modul vs. Masa)

* **Zaključak:** Odabran je modul  pri osnom razmaku.
* **Inženjersko opravdanje:** Modul 1.5 je "zlatna sredina" za snagu od 250W. Manji modul bi bio previše krhak, dok bi veći (npr. ) nepotrebno povećao gabarite i težinu reduktora, što bi ga učinilo skupljim za proizvodnju i težim za ugradnju.

### 3. Kvaliteta Zahvata (Faktor Prekrivanja )

* **Zaključak:** Izračunata vrijednost .
* **Inženjersko opravdanje:** Brojka 1.63 govori da su zubi u zahvatu "preklopljeni" 63% vremena. To znači da dok jedan par zubi izlazi iz zahvata, drugi je već duboko unutra, što drastično smanjuje **dinamičke udare** i buku pri 1800 RPM.

### 4. Sigurnost na Lom (Napon u korijenu )

* **Zaključak:** Napon  je značajno niži od dopuštenih  za čelik C45.
* **Inženjersko opravdanje:** Visok faktor sigurnosti () namjerno je ostavljen kako bi se kompenzirao **startni moment motora**. Elektromotori pri kretanju mogu povući i do 2-3 puta veću struju (i moment), a dizajn garantira da zubi to mogu izdržati bez deformacije.

### 5. Sigurnost na Trošenje (Dodirni napon )

* **Zaključak:** Hertzov tlak  je unutar granica izdržljivosti materijala.
* **Inženjersko opravdanje:** Ovo je ključ za dugovječnost. Budući da je površinski pritisak kontroliran, izbjegnuta je pojava **pittinga**. Čak i uz dugotrajni rad (npr. 8 sati dnevno), bokovi zubi će zadržati svoj geometrijski oblik, što sprječava pojavu "zvijanja" i vibracija tijekom vremena.

### 6. Mehanička Sila i Vratila

* **Zaključak:** Izračunata vrijednost
* **Inženjersko opravdanje:** Ova sila je mjerodavna za proračun ležajeva i promjer vratila. S obzirom na to da je  relativno mala, omogućuje korištenje standardnih kugličnih ležajeva, što dodatno pojeftinjuje cijeli sklop transportera.
