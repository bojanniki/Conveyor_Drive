# Conveyor Drive Project

---

## 1. Uvod i opis projekta

Ovaj projekt obuhvaća konstrukciju i dokumentaciju jednostupanjskog 
zupčanog reduktora namijenjenog pogonu transportne trake.

### 1.1. Zadatak projekta

Projektni zadatak definira izradu kompaktnog reduktora s paralelnim 
osovinama koji transformira visokobrzinsko gibanje standardnog AC motora 
u sporije, moment-bogatije gibanje pogodnog za pogon transportera.

### 1.2. Opseg dokumentacije

Sukladno zadatku, projekt obuhvaća:
- **3D sklop** – kompletan model u Solid Edgeu s montažnim vezama
- **2D nacrti** – nacrti svih nestandardnih dijelova s tolerancijama
- **Tablica zupčanika** – prema DIN 3966 za oba zupčanika
- **Popis dijelova (BOM)** – svih 18 komponenti sklopa
- **Tehnički proračun** – validacija prema ISO standardima

### 1.3. Konačni dizajn

Inicijalni proračun rezultirao je nestandardnim modulom m = 1.38, 
koji je odbačen u korist standardnog **m = 1.5**. Konačni dizajn 
koristi standardizirani par zupčanika **z₁ = 18 / z₂ = 72**, 
što osigurava točan prijenosni omjer **4:1** i osni razmak **67.5 mm**.

Kućište je izvedeno od sivog lijeva **EN-GJL-250**, a zupčanici i 
vratila od čelika pogodnog za toplinsku obradu. Brtvljenje je 
riješeno brtvenim papirom na naležnim površinama, a podmazivanje 
centralnom mazalicom.

### 2.2. Glavne inženjerske veličine i definicije

Prilikom proračuna zupčanog prijenosnika, ključno je razumijevanje 
sljedećih parametara koji izravno utječu na rad i vijek trajanja sustava:

* **Modul ($m$):** Osnovna mjera veličine zuba i najvažniji standardizirani parametar. Definira se kao odnos diobenog promjera i broja zubi ($m = d/z$). Svi zupčanici u zahvatu moraju imati identičan modul.
* **Diobeni promjer ($d$):** Teoretski promjer na kojem se dva zupčanika kotrljaju jedan po drugom bez klizanja. To je referentna linija za sve ostale geometrijske izračune.
* **Osni razmak ($a$):** Najkraća udaljenost između središta vratila pogonskog i gonjenog zupčanika. Ključan je za precizno projektiranje kućišta reduktora.
* **Faktor prekrivanja ($\epsilon_\alpha$):** Broj koji pokazuje prosječan broj parova zubi koji su istovremeno u dodiru. Za miran i tih rad bez udara, ova vrijednost bi trebala biti veća od $1.2$.
* **Napon u korijenu zuba ($\sigma_F$):** Kritično naprezanje koje nastaje na bazi zuba uslijed djelovanja obodne sile. Ako ovaj napon prijeđe dopuštene granice, dolazi do loma zuba.
* **Hertzov dodirni tlak ($\sigma_H$):** Tlak koji nastaje na dodirnim površinama (bokovima) zuba. On je mjerodavan za pojavu **pittinga** (zamora materijala na površini boka).
* **Faktori sigurnosti ($S_F, S_H$):** Odnos između maksimalno izdržljivog naprezanja materijala i stvarnog radnog naprezanja. U strojarstvu se cilja na vrijednosti iznad $1.2$ kako bi se osigurala pouzdanost u nepredviđenim uvjetima rada.

## 3. Tehnički proračun prijenosnog sustava

Ovaj odjeljak sadrži matematičku analizu i validaciju zupčanog prijenosnika 
prema ISO standardima, usklađenu s rezultatima dobivenim u 
*Solid Edge Engineering Reference* modulu.

---

### 3.1. Ulazni parametri (Specifikacije pogona)

Ovi podaci definiraju radne uvjete stroja.

| Varijabla | Naziv | Vrijednost | Opis |
| :--- | :--- | :--- | :--- |
| $P$ | **Snaga** | **250 W** | Nazivna snaga pogonskog motora. |
| $n_1$ | **Ulazna brzina** | **1800 RPM** | Brzina vrtnje pogonskog vratila. |
| $n_2$ | **Izlazna brzina** | **450 RPM** | Projektirana brzina na izlazu reduktora. |
| $\eta$ | **Efikasnost** | **0.98** | Faktor gubitaka (zahvat i ležajevi). |
| $\alpha$ | **Kut zahvata** | **20°** | Standardni kut pritiska profila zuba. |

---

### 3.2. Kinematika i momenti

Transformacija gibanja i sila unutar prijenosnika za standardizirani par $18/72$.

* **Prijenosni omjer ($i$):**
  $$i = \frac{z_2}{z_1} = \frac{72}{18} = 4.0$$
* **Kutna brzina ($\omega_1$):**
  $$\omega_1 = \frac{2 \cdot \pi \cdot 1800}{60} = 188.5 \text{ rad/s}$$
* **Okretni moment na pinion-u ($T_1$):**
  $$T_1 = \frac{P}{\omega_1} = \frac{250}{188.5} = 1.33 \text{ Nm}$$

---

### 3.3. Geometrija zupčanika

Konačne dimenzije prilagođene standardnom modulu $m=1.5$.

* **Modul ($m$):** **1.5 mm** (Standardni profil)
* **Diobeni promjeri ($d$):**
  $$d_1 = m \cdot z_1 = 27.0 \text{ mm}$$
  $$d_2 = m \cdot z_2 = 108.0 \text{ mm}$$
* **Osni razmak ($a$):**
  $$a = \frac{d_1 + d_2}{2} = 67.5 \text{ mm}$$
* **Obodna brzina ($v$):**
  $$v = \frac{d_1 \cdot \pi \cdot n_1}{60000} = 2.54 \text{ m/s}$$

---

### 3.4. Faktor prekrivanja ($\epsilon_\alpha$)

Faktor prekrivanja pokazuje koliko parova zubi je istovremeno 
u zahvatu. Za miran i tih rad potrebno je **ε_α > 1.2**.

#### Formula:

$$\epsilon_\alpha = \frac{1}{2\pi} \left[ z_1 \cdot (\tan\alpha_{a1} - \tan\alpha') + z_2 \cdot (\tan\alpha_{a2} - \tan\alpha') \right]$$

Gdje su:
- **α' = 20°** – kut zahvata (standardni profil)
- **α_a1, α_a2** – kutovi tjemenog zahvata piniona i zupčanika

#### Kutovi tjemenog zahvata:

$$\alpha_{a1} = \arccos\left(\frac{r_1 \cdot \cos\alpha'}{r_{a1}}\right) = 32.25°$$

$$\alpha_{a2} = \arccos\left(\frac{r_2 \cdot \cos\alpha'}{r_{a2}}\right) = 23.89°$$

#### Izračun:

$$\epsilon_\alpha = \frac{1}{2\pi} \left[ 18 \cdot (\tan 32.25° - \tan 20°) + 72 \cdot (\tan 23.89° - \tan 20°) \right]$$

$$\epsilon_\alpha \approx 1.67$$

#### Zaključak:

Faktor prekrivanja **ε_α = 1.67** znači da su u prosjeku 
**1.67 para zubi** istovremeno u zahvatu, što je iznad 
minimuma **1.2** i osigurava:
- ✅ Miran i tih rad bez udara
- ✅ Ravnomjerni prijenos opterećenja
- ✅ Duži vijek trajanja zuba

---

### 3.5. Sile u zahvatu

Mehanička opterećenja dobivena iz momenta $T_1$.

* **Obodna sila ($F_t$):**
  $$F_t = \frac{2000 \cdot T_1}{d_1} = 98.5 \text{ N}$$
* **Radijalna sila ($F_r$):**
  $$F_r = F_t \cdot \tan(20^\circ) = 35.8 \text{ N}$$

---

### 3.6. Proračun vratila i odabir promjera

Na temelju izlaznog momenta $T_2 = T_1 \cdot i \cdot \eta \approx 5.3 \text{ Nm}$ 
i sila u zahvatu, određuje se minimalni promjer.

* **Dopušteno naprezanje ($\tau_{t, dop}$):** **40 MPa** (Materijal C45 prema Deckerovoj tablici za vratila prijenosnika).
* **Teoretski minimalni promjer ($d_{min}$):**
  $$d_{min} = \sqrt[3]{\frac{T_2}{0.2 \cdot \tau_{t, dop}}} = \sqrt[3]{\frac{5300 \text{ Nmm}}{0.2 \cdot 40}} \approx 8.72 \text{ mm}$$
* **Konstrukcijski odabir:** Usvaja se jedinstveni promjer od **12.0 mm** za oba vratila.
  * **Obrazloženje:** Kompenzacija slabljenja presjeka zbog utora za pero (DIN 6885) i unifikacija s ležajevima serije **6201**.

---

### 3.7. Proračun i validacija veze vratilo-glavina (Pero)

Za prijenos okretnog momenta s vratila na zupčanike odabrana je veza perom 
prema standardu **DIN 6885**. Proračun se temelji na provjeri površinskog 
pritiska na bokovima pera prema preporukama iz priručnika **Decker**.

#### 3.7.1. Geometrija i odabir komponente

Na temelju usvojenog promjera vratila $d = 12 \text{ mm}$, odabrane su standardne dimenzije pera:

| Oznaka | Opis | Vrijednost [mm] | Izvor |
| :--- | :--- | :--- | :--- |
| $b$ | Širina pera | 4 | DIN 6885 |
| $h$ | Visina pera | 4 | DIN 6885 |
| $L$ | Ukupna duljina | 12 | Odabrano (širina zupčanika 15 mm) |
| $t_1$ | Dubina utora u vratilu | 2.5 | DIN 6885 |
| $t_2$ | Dubina utora u zupčaniku | 1.8 | DIN 6885 |

#### 3.7.2. Parametri proračuna

Proračun se provodi za kritično, izlazno vratilo gdje je opterećenje najveće.

* **Okretni moment ($T$):** $5.3 \text{ Nm} = 5300 \text{ Nmm}$
* **Promjer vratila ($d$):** $12 \text{ mm}$
* **Aktivna visina boka ($h - t_1$):** $4 \text{ mm} - 2.5 \text{ mm} = 1.5 \text{ mm}$
* **Nosiva duljina ($l_t$):** $L - b = 12 \text{ mm} - 4 \text{ mm} = 8 \text{ mm}$ (za pero oblika A)
* **Dopušteni pritisak ($p_{dop}$):** $\approx 100 \text{ MPa}$ (za materijal C45 prema Deckeru)

#### 3.7.3. Kontrola površinskog pritiska

$$p = \frac{2 \cdot T}{d \cdot (h - t_1) \cdot l_t} = \frac{2 \cdot 5300}{12 \cdot 1.5 \cdot 8} \approx \mathbf{73.61 \text{ MPa}}$$

#### 3.7.4. Zaključak

Budući da je $p = 73.61 \text{ MPa} < p_{dop} = 100 \text{ MPa}$:

1. **Odabrano pero DIN 6885 - A 4 x 4 x 12 u potpunosti zadovoljava uvjete čvrstoće.**
2. **Postoji rezerva nosivosti od cca 26%, što osigurava stabilan rad pri dinamičkim opterećenjima.**
3. **Dimenzije utora $t_1=2.5 \text{ mm}$ i $t_2=1.8 \text{ mm}$ korištene su pri modeliranju u Solid Edgeu.**

### 3.8. Odabir i proračun ležajeva

#### 3.8.1. Konfiguracija ležajeva

Reduktor koristi **4 ležaja** raspoređena simetrično:
- 2 ležaja na ulaznoj osovini (pinion, $n_1$ = 1800 RPM)
- 2 ležaja na izlaznoj osovini (gear, $n_2$ = 450 RPM)

Razmak između ležajeva na istoj osovini: **L = 67.5 mm**  
Zupčanici su pozicionirani **centralno između ležajeva** (simetrično).

#### 3.8.2. Sile na ležajeve

Budući da je zupčanik centralno postavljen, opterećenje se dijeli jednako na oba ležaja iste osovine:

$$F_{res} = \sqrt{F_t^2 + F_r^2} = \sqrt{98.5^2 + 35.8^2} \approx 104.8 \text{ N}$$

$$F_{ležaj} = \frac{F_{res}}{2} \approx 52.4 \text{ N}$$

#### 3.8.3. Odabir ležaja – 6201-2RS

Na temelju promjera vratila **d = 12 mm**, odabran je standardni kuglični ležaj:

| Parametar | Vrijednost |
|:----------|:-----------|
| Oznaka | 6201-2RS |
| Unutarnji promjer | 12 mm |
| Vanjski promjer | 32 mm |
| Širina | 10 mm |
| Dinamička nosivost (C) | 6.82 kN |
| Statička nosivost (C₀) | 3.05 kN |
| Brtvljenje | 2RS (s brtvama) |

**Obrazloženje odabira 2RS:** Ležaj s gumenim brtvama s obje strane dolazi tvornički 
namazan i zatvoren – idealno za mazanje centralnom mazalicom bez dodatnih brtvi na osovinama.

#### 3.8.4. Vijek trajanja ležaja (L10)

Proračun prema **ISO 281**:

$$L_{10} = \left(\frac{C}{F}\right)^3 \cdot \frac{10^6}{60 \cdot n}$$

**Kritični ležaj – ulazna osovina (1800 RPM):**
$$L_{10} = \left(\frac{6820}{52.4}\right)^3 \cdot \frac{10^6}{60 \cdot 1800} \approx 227{,}000 \text{ sati}$$

**Ležaj izlazne osovine (450 RPM):**
$$L_{10} = \left(\frac{6820}{52.4}\right)^3 \cdot \frac{10^6}{60 \cdot 450} \approx 908{,}000 \text{ sati}$$

#### 3.8.5. Zaključak

Izračunati vijek trajanja daleko premašuje standardnih **20,000 sati** za industrijsku primjenu. Ležaj **6201-2RS** je konzervativno predimenzioniran za ovo opterećenje, što je svjesna odluka zbog:
- Niske cijene standardnog ležaja
- Jednostavne nabave i zamjene
- Dodatne sigurnosti pri startnim momentima transportera

---

### 3.9. Analiza čvrstoće i faktora sigurnosti

Usporedba stvarnih naprezanja s dopuštenim granicama za **čelik C45 (normaliziran)**.

#### A. Sigurnost na lom zuba ($S_F \approx 14.1$)

Izračunato naprezanje u korijenu iznosi **14.18 N/mm²**.
* **Analiza:** Budući da je dopušteno naprezanje za C45 $\sigma_{F, dop} = 200 \text{ N/mm²}$, faktor sigurnosti je izuzetno visok.
* **Zaključak:** Dizajn je imun na zamorni lom korijena zuba pri nazivnom opterećenju i startnim momentima.

#### B. Sigurnost na pitting ($S_H \approx 2.1$)

Hertzov dodirni tlak na bokovima zuba iznosi **260.72 N/mm²**.
* **Analiza:** Napon je ispod granice izdržljivosti materijala ($\sigma_{H, dop} = 550 \text{ N/mm²}$).
* **Mjerodavnost:** Faktor **2.1** potvrđuje da će sustav imati dug radni vijek bez ljuštenja površine zuba, što je kritično za preciznost rada transportera.

## 6. Konačna ocjena dizajna

Standardizirani dizajn sustava ($m=1.5, z=18/72$) ocijenjen je kao **visoko pouzdan**.

### 6.1. Inženjerska iteracija i standardizacija

Tijekom procesa projektiranja provedena je ključna inženjerska optimizacija. 
Inicijalni proračuni rezultirali su nestandardnim **modulom $m = 1.38$**. 
Iako teoretski ispravan, taj model je odbačen u korist **standardnog modula $m = 1.5$**.

Ovim prijelazom postignuta je:
* **Ekonomičnost:** Korištenje kataloških (*off-the-shelf*) zupčanika značajno smanjuje troškove i vrijeme izrade.
* **Održivost:** Osigurana je brza i jednostavna zamjenjivost dijelova u slučaju servisiranja.
* **Geometrijska točnost:** Zadržan je projektirani osni razmak od **67.5 mm** uz precizan prijenosni omjer **4:1** ($z_1 = 18, z_2 = 72$).

### 6.2. Statička i dinamička sigurnost

Analiza naprezanja u korijenu zuba potvrdila je visoku sigurnost na lom. 
Za odabrani materijal **čelik C45 (normaliziran)**, izračunato naprezanje 
je znatno ispod dopuštenih $200 \text{ N/mm²}$. Faktor sigurnosti na 
savijanje ($S_F \approx 14.1$) jamči otpornost na iznenadna preopterećenja.

### 6.3. Otpornost na trošenje (Pitting)

Faktor sigurnosti na dodirni pritisak **$S_H \approx 2.1$** potvrđuje 
stabilnost sustava protiv površinskog zamora materijala (pittinga), 
što je ključno za dugovječan rad pri brzini od 1800 RPM.

### 6.4. Funkcionalnost i točnost

Odabrana **ISO 8 klasa točnosti** osigurava optimalan balans između 
cijene izrade i kvalitete rada. Faktor prekrivanja $\epsilon_\alpha = 1.67$ 
osigurava kontinuiran zahvat i miran tihi rad.

**Finalna ocjena:** Projektirani zupčani prijenosnik u potpunosti zadovoljava 
tehničke, ekonomske i sigurnosne kriterije.

## Zaključak

Na temelju provedenih proračuna u programskom jeziku **Python** te validacije 
unutar **Solid Edge Engineering Reference** modula, izvedeni su sljedeći 
zaključci o mehaničkom integritetu sustava:

Projektirani zupčani prijenosnik u potpunosti zadovoljava tehničke, ekonomske 
i sigurnosne kriterije. Prijelaz s inicijalno izračunatog modula $1.38$ na 
standardni modul $1.5$ predstavlja primjer ispravne inženjerske prakse i 
optimizacije resursa u razvoju mehaničkih sustava.

| Provjera | Rezultat | Status |
|:---------|:---------|:-------|
| Prijenosni omjer | 4:1 | ✅ |
| Faktor prekrivanja $\epsilon_\alpha$ | 1.67 > 1.2 | ✅ |
| Sigurnost na lom $S_F$ | 14.1 > 1.2 | ✅ |
| Sigurnost na pitting $S_H$ | 2.11 > 1.2 | ✅ |
| Sigurnost vratila | 2.66 > 1.2 | ✅ |
| Sigurnost pera | 1.38 > 1.2 | ✅ |
| Vijek ležaja $L_{10}$ | 227,000 h | ✅ |

````python
import math

# =================================================================
# 1. ULAZNI PODACI (za standardizirane zupčanike)
# =================================================================
P = 0.25             # Snaga [kW] (promijenjeno u kW radi lakšeg računa)
n1 = 1800            # Ulazna brzina [RPM]
eta = 0.98           # Poboljšana efikasnost prema Solid Edge izvještaju
alpha_deg = 20       # Kut zahvata [stupnjevi]

# Geometrijski izbor (STANDARDNI MODUL I PAR)
m = 1.5              # Standardni modul [mm]
z1 = 18              # Broj zuba malog zupčanika (piniona)
z2 = 72              # Broj zuba velikog zupčanika (i=4)
b = 15.0             # Širina zupčanika [mm]

# Materijal i dopuštena naprezanja (C45 - Normaliziran)
Yf = 0.308           # Precizniji Lewisov faktor oblika za z=18
ZE = 189.8           # Faktor elastičnosti prema SE Reportu [sqrt(N/mm2)]
sigma_F_dop = 200    # Dopušteni napon savijanja [N/mm2]
sigma_H_dop = 550    # Dopušteni dodirni tlak [N/mm2]

# =================================================================
# 2. KINEMATIKA, MOMENTI I GEOMETRIJA
# =================================================================
i = z2 / z1
n2 = n1 / i
T1 = (P * 1000) / (2 * math.pi * n1 / 60)  # Moment na pinion-u [Nm]
T2 = T1 * i * eta                          # Moment na gear-u [Nm]

d1 = m * z1                                # Diobeni promjer 1 [mm]
d2 = m * z2                                # Diobeni promjer 2 [mm]
a = (d1 + d2) / 2                          # Standardni osni razmak [mm]
v = (d1 * math.pi * n1) / 60000            # Obodna brzina [m/s]

# Sile
Ft = (2000 * T1) / d1                      # Obodna sila [N]
alpha_rad = math.radians(alpha_deg)
Fr = Ft * math.tan(alpha_rad)              # Radijalna sila [N]

# =================================================================
# 2a. FAKTOR PREKRIVANJA (ispravljena formula)
# =================================================================
r1 = d1 / 2      # Polumjer diobene kružnice piniona
r2 = d2 / 2      # Polumjer diobene kružnice zupčanika
d_a1 = d1 + 2 * m    # Tjemeni promjer piniona [mm]
d_a2 = d2 + 2 * m    # Tjemeni promjer zupčanika [mm]
ra1 = d_a1 / 2   # Polumjer tjemene kružnice piniona
ra2 = d_a2 / 2   # Polumjer tjemene kružnice zupčanika

alpha_a1 = math.acos(r1 * math.cos(alpha_rad) / ra1)
alpha_a2 = math.acos(r2 * math.cos(alpha_rad) / ra2)

epsilon_alpha = (1 / (2 * math.pi)) * (
    z1 * (math.tan(alpha_a1) - math.tan(alpha_rad)) +
    z2 * (math.tan(alpha_a2) - math.tan(alpha_rad))
)

# =================================================================
# 3. PRORAČUN NAPONA I SIGURNOSTI
# =================================================================
# Napon u korijenu (Savijanje)
sigma_F = Ft / (b * m * Yf)

# Dodirni napon (Hertzov tlak)
# ZH - faktor zahvata (geometrijski faktor boka)
ZH = math.sqrt(2 / (math.sin(alpha_rad) * math.cos(alpha_rad)))
sigma_H = ZE * ZH * math.sqrt((Ft / (b * d1)) * ((i + 1) / i))

# Faktori sigurnosti (SF i SH)
SF = sigma_F_dop / sigma_F
SH = sigma_H_dop / sigma_H

# =================================================================
# 3a. PRORAČUN VRATILA (Osovina)
# =================================================================
tau_t_dop = 40          # Dopušteno naprezanje na uvijanje [N/mm2] (iz tablice)
d_vratila = 12.0        # Odabrani standardni promjer [mm]

# Teoretski minimalni promjer izlaznog vratila (Deckerova formula)
# Koristi se T2 jer je moment na izlazu najveći
d_min = (T2 * 1000 / (0.2 * tau_t_dop))**(1/3)

# Stvarno naprezanje na uvijanje u odabranom vratilu
tau_t = (T2 * 1000) / (0.2 * d_vratila**3)
S_vratilo = tau_t_dop / tau_t  # Faktor sigurnosti vratila

# =================================================================
# 3b. PRORAČUN PERA (DIN 6885)
# =================================================================
# Dimenzije pera za d=12mm
b_pero = 4.0            # Širina [mm]
h_pero = 4.0            # Visina [mm]
L_pero = 12.0           # Ukupna duljina [mm]
t1 = 2.5                # Dubina utora u vratilu [mm]
p_dop = 100.0           # Dopušteni površinski pritisak [N/mm2]

# Proračun nosive duljine i aktivne visine
l_t = L_pero - b_pero   # Nosiva duljina (Oblik A)
h_aktivno = h_pero - t1 # Visina boka koja dodiruje zupčanik

# Površinski pritisak na boku pera (Prilagođena Decker formula)
p_pero = (2 * T2 * 1000) / (d_vratila * h_aktivno * l_t)
S_pero = p_dop / p_pero # Faktor sigurnosti pera

# =================================================================
# 4. INŽENJERSKI IZVJEŠTAJ (Standardiziran)
# =================================================================
print(f"{'FINALNI STANDARDIZIRANI PRORAČUN ':=^60}")
print(f"Prijenosni omjer (i):           {i:.2f} (Standardni 4:1)")
print(f"Osni razmak (a):                {a:.2f} mm (Standardni)")
print(f"Modul (m):                      {m:.2f} mm")
print(f"Obodna sila (Ft):               {Ft:.2f} N")
print("-" * 60)
print(f"{'FAKTOR PREKRIVANJA ':=^60}")
print(f"Tjemeni kut piniona  (α_a1):    {math.degrees(alpha_a1):.2f}°")
print(f"Tjemeni kut zupčanika (α_a2):   {math.degrees(alpha_a2):.2f}°")
print(f"Faktor prekrivanja   (ε_α):     {epsilon_alpha:.2f}")
print("-" * 60)

if epsilon_alpha > 1.2:
    print(f"STATUS: ε_α = {epsilon_alpha:.2f} > 1.2 – Zahvat je MIRAN i TIH ✓")
else:
    print(f"UPOZORENJE: ε_α = {epsilon_alpha:.2f} < 1.2 – Mogući udari u zahvatu!")
print("=" * 60)
print(f"BROJ ZUBA:                      z1 = {z1}, z2 = {z2}")
print(f"MOMENTI:                        T1 = {T1:.2f} Nm, T2 = {T2:.2f} Nm")
print(f"OBODNA BRZINA (v):              {v:.2f} m/s")
print("-" * 60)
print(f"{'PRORAČUN VRATILA I PERA ':=^60}")
print(f"Minimalni promjer (d_min):      {d_min:.2f} mm")
print(f"Odabrani promjer (d_vratila):   {d_vratila:.2f} mm")
print(f"Sigurnost vratila (S_vratilo):  {S_vratilo:.2f}")
print("-" * 60)
print(f"Površinski pritisak pera (p):   {p_pero:.2f} N/mm2")
print(f"Dopušteni pritisak (p_dop):     {p_dop:.2f} N/mm2")
print(f"Sigurnost pera (S_pero):        {S_pero:.2f}")
print("=" * 60)
print(f"NAPON SAVIJANJA (sigma_F):      {sigma_F:.2f} N/mm2")
print(f"SIGURNOST NA LOM (SF):          {SF:.2f} (Dopušteno: {sigma_F_dop} N/mm2)")
print("-" * 60)
print(f"DODIRNI NAPON (sigma_H):        {sigma_H:.2f} N/mm2")
print(f"SIGURNOST NA PITTING (SH):      {SH:.2f} (Dopušteno: {sigma_H_dop} N/mm2)")
print("=" * 60)

# Zaključna poruka
if SF > 1.2 and SH > 1.2:
    print("STATUS: Dizajn je VALIDIRAN i STANDARDIZIRAN. Komponente su sigurne.")
````
