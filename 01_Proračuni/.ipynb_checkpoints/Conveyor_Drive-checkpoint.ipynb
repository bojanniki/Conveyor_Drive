# Conveyor Drive Project

## Dizajn i proračun malog reduktora za laboratorijsku traku

### Zadane veličine:


* **Snaga (P):** 250 W
* **Ulazna brzina ($n_1$):** 1800 RPM
* **Izlazna brzina ($n_2$):** 450 RPM (4:1 Redukcija)
* **Konfiguracija:** Paralelne osovine

## Formule za izračun

### Kinematičke veličine

### Prijenosni omjer ($i$)

Prijenosni omjer predstavlja omjer brzina vrtnje ulaznog i izlaznog vratila, što je ujedno jednako omjeru broja zuba gonjenog i pogonskog zupčanika:

$$i = \frac{n_1}{n_2} = \frac{z_2}{z_1}$$

**Značenje:**
* $i > 1$: Reduktor (smanjuje brzinu, povećava moment)
* $i < 1$: Multiplikator (povećava brzinu, smanjuje moment).

### Kutna brzina ($\omega$)

Kutna brzina definira brzinu promjene kuta rotacije u jedinici vremena. Za pogonsko vratilo računa se prema formuli:

$$\omega_1 = \frac{2 \cdot \pi \cdot n_1}{60} \quad [\text{rad/s}]$$

**Gdje su:**
* $n_1$ – brzina vrtnje vratila [RPM]
* $\pi$ – konstanta Pi ($\approx 3.14159$)
* $60$ – broj sekundi u minuti

### Odabir broja zubi ($z$)

Broj zubi pogonskog zupčanika ($z_1$) bira se tako da se izbjegne podrezivanje korijena zuba i osigura miran rad. 

**Kriteriji za odabir:**
* **Granični broj zubi:** Za standardni kut zahvata $\alpha = 20^\circ$, minimalni broj zubi iznosi $z_{min} = 17$. 
* **Prijenosni omjer:** Broj zubi velikog zupčanika računa se kao $z_2 = z_1 \cdot i$.
* **Preporuka za projekt:** Odabiremo $z_1 = 18$ kako bismo dobili kompaktan prijenosnik uz siguran zahvat bez podrezivanja.

$$z_2 = z_1 \cdot i$$

### Odabir modula ($m$)

Modul je osnovna mjera veličine zuba zupčanika. Svi ostali geometrijski parametri zupčanika izvode se iz modula. Definira se kao:

$$m = \frac{d}{z} \quad [\text{mm}]$$

**Kriteriji za odabir:**
* **Čvrstoća:** Veći modul znači deblji korijen zuba i veću otpornost na lom.
* **Standardizacija:** Modul se uvijek bira iz niza standardnih brojeva (npr. ISO standard) kako bi se mogli koristiti standardni alati za izradu.
* **Preporuka za projekt:** S obzirom na snagu od 250 W, preporučeni standardni moduli su obično u rasponu od **1.5 do 2.5 mm**. Odabrano **1,5 mm**.

### Diobeni promjer ($d$)

Diobeni promjer je promjer zamišljene kružnice (diobene kružnice) na kojoj je debljina zuba jednaka širini međuzublja. Računa se kao umnožak modula i broja zubi:

$$d = m \cdot z \quad [\text{mm}]$$

**U ovom projektu imamo dva promjera:**
* $d_1 = m \cdot z_1$ – Promjer pogonskog zupčanika (piniona)
* $d_2 = m \cdot z_2$ – Promjer gonjenog zupčanika (velikog zupčanika)

Ove vrijednosti su ključne za određivanje **osnog razmaka** i **brzine vrtnje**.

### Tjemeni promjer ($d_a$)

Tjemeni promjer (engl. *Addendum Diameter*) definira vanjski rub zupčanika. Dobiva se dodavanjem visine glave zuba (koja je kod standardnih zupčanika jednaka modulu $m$) na diobeni promjer:

$$d_a = d + 2 \cdot m \quad [\text{mm}]$$

**Za projekt:**
* $d_{a1} = d_1 + 2 \cdot m$ – Vanjski promjer malog zupčanika
* $d_{a2} = d_2 + 2 \cdot m$ – Vanjski promjer velikog zupčanika

Ova mjera je ključna za određivanje **unutarnjih dimenzija kućišta** reduktora kako zupčanici ne bi strugali o stijenke.

### Podnožni promjer ($d_f$)

Podnožni promjer (engl. *Root Diameter*) je promjer kružnice koja spaja dna međuzublja. Izračunava se oduzimanjem visine korijena zuba od diobenog promjera. Standardni faktor za visinu korijena je $1.25 \cdot m$ sa svake strane (ukupno $2.5 \cdot m$):

$$d_f = d - 2.5 \cdot m \quad [\text{mm}]$$

**Značenje za projekt:**
* Definira najslabiju točku zupčanika (napon u korijenu zuba).
* Razlika između $d_a$ i $d_f$ daje nam punu visinu zuba ($h = 2.25 \cdot m$).
* Između tjemena jednog zuba i podnožja drugog ostaje zračnost $c = 0.25 \cdot m$ kako zubi ne bi zapinjali u dnu.

### Osni razmak ($a$) i Širina zupčanika ($b$)

**Osni razmak** je udaljenost između središta dvaju vratila. Kod zupčanika bez pomaka profila, on je jednak poluzbroju diobenih promjera:
$$a = \frac{d_1 + d_2}{2} = m \cdot \frac{z_1 + z_2}{2} \quad [\text{mm}]$$

**Širina zupčanika** ($b$) određuje se na temelju faktora širine $\psi_d$, koji za standardne reduktore obično iznosi između $0.6$ i $1.2$. Računa se prema diobenom promjeru manjeg zupčanika:
$$b = \psi_d \cdot d_1 \quad [\text{mm}]$$

*Veća širina $b$ znači manji pritisak na bokove zubi, ali zahtijeva preciznije poravnanje osovina.*

### Dinamičke veličine i sile

Sile koje djeluju u zahvatu zupčanika ključne su za odabir ležajeva i provjeru čvrstoće materijala.

**Izlazni moment ($T_2$):**
Povećava se s prijenosnim omjerom, ali se smanjuje zbog gubitaka:
$$T_2 = T_1 \cdot i \cdot \eta \quad [\text{Nm}]$$

**Obodna sila ($F_t$):**
Glavna sila koja vrši koristan rad (okretanje):
$$F_t = \frac{2 \cdot T}{d} \quad [\text{N}]$$

**Radijalna sila ($F_r$):**
Sila koja pritišće vratila i opterećuje ležajeve, pokušavajući razdvojiti zupčanike:
$$F_r = F_t \cdot \tan(\alpha) \quad [\text{N}]$$
*(Standardni kut zahvata $\alpha = 20^\circ$)*

### Parametri pouzdanosti

Da bi reduktor bio dugotrajan i tih, moraju biti zadovoljeni sljedeći kriteriji:

1. **Faktor prekrivanja ($\epsilon$):** Mora biti $\epsilon > 1.2$. To jamči da novi par zubi uđe u zahvat prije nego što prethodni izađe (miran rad).
2. **Minimalni broj zubi ($z_{min}$):** Da bi se izbjeglo **podrezivanje** (oslabljivanje korijena zuba), preporučuje se $z \ge 17$.
3. **Napon u korijenu ($\sigma_F$):** Osigurava da sila $F_t$ ne slomi zub kao konzolu.
4. **Dodirni napon ($\sigma_H$):** Osigurava da površina zuba ne popuca uslijed pritiska (pitting).

```python
import math

# ==========================================
# 1. ULAZNI PODACI (Zadane veličine)
# ==========================================
P = 250              # Snaga [W]
n1 = 1800            # Ulazna brzina [RPM]
n2 = 450             # Izlazna brzina [RPM]
eta = 0.97           # Efikasnost (gubici trenja)
alpha = 20           # Kut zahvata [stupnjevi]

# Geometrijski izbor
m = 1.5              # Modul [mm]
z1 = 18              # Broj zuba malog zupčanika

# ==========================================
# 2. PRORAČUN KINEMATIKE I MOMENATA
# ==========================================
# Prijenosni omjer
i = n1 / n2

# Kutne brzine [rad/s]
omega1 = (2 * math.pi * n1) / 60
omega2 = (2 * math.pi * n2) / 60

# Okretni momenti [Nm]
T1 = P / omega1
T2 = T1 * i * eta

# ==========================================
# 3. PRORAČUN GEOMETRIJE ZUPČANIKA
# ==========================================
# Broj zuba velikog zupčanika
z2 = int(z1 * i)

# Diobeni promjeri [mm]
d1 = m * z1
d2 = m * z2

# Tjemeni promjeri [mm]
da1 = d1 + 2 * m
da2 = d2 + 2 * m

# Podnožni promjeri [mm]
df1 = d1 - 2.5 * m
df2 = d2 - 2.5 * m

# Osni razmak [mm]
a = (d1 + d2) / 2

# Obodna brzina [m/s]
v = (d1 * math.pi * n1) / 60000

# ==========================================
# 4. PRORAČUN SILA I ŠIRINE
# ==========================================
# Obodna sila [N] (Moment u Nmm / polumjer u mm)
Ft = (2000 * T1) / d1

# Radijalna sila [N]
Fr = Ft * math.tan(math.radians(alpha))

# Preporučena širina zupčanika [mm] (psi_d = 0.8)
b = 0.6 * d1

# ==========================================
# 5. ISPIS REZULTATA (Inženjerski izvještaj)
# ==========================================
print(f"{' REZULTATI PRORAČUNA REDUKTORA ':=^40}")
print(f"Prijenosni omjer (i):      {i:.2f}")
print(f"Kutne brzina (omega1):      {omega1:.2f}")
print(f"Kutne brzina (omega2):      {omega2:.2f}")
print(f"Izlazni moment (T2):       {T2:.2f} Nm")
print("-" * 40)
print(f"BROJ ZUBA:                 z1 = {z1}, z2 = {z2}")
print(f"DIOBENI PROMJERI:          d1 = {d1:.1f} mm, d2 = {d2:.1f} mm")
print(f"TJEMENI PROMJERI (Vanjski): da1 = {da1:.1f} mm, da2 = {da2:.1f} mm")
print(f"PODNOZNI PROMJERI (Vanjski): df1 = {df1:.1f} mm, df2 = {df2:.1f} mm")
print(f"OSNI RAZMAK (a):           {a:.2f} mm")
print(f"ŠIRINA ZUPČANIKA (b):      {b:.1f} mm")
print("-" * 40)
print(f"OBODNA SILA (Ft):          {Ft:.2f} N")
print(f"RADIJALNA SILA (Fr):       {Fr:.2f} N")
print(f"OBODNA BRZINA (v):         {v:.2f} m/s")
print("=" * 40)

# Brza provjera za podrezivanje
if z1 < 17:
    print("NAPOMENA: Postoji rizik od podrezivanja korijena zuba!")
```
